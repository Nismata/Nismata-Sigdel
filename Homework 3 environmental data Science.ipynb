```python
# Import packages
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# read the file
fileName = r"c:\users\user\Downloads\LA_AQS_2023.csv" #note on path
pd.read_csv(fileName)
```

```python
df = pd.read_csv(fileName) #store file in variable
```

```python
df.head() #file describe
# GOTO Filtering
```

```python
df['Parameter Name'].unique()
```

```python
df_O3 = df[
    (df['Parameter Name'] == 'Ozone') &
    (df['Duration Description'] == '1 HOUR')
]
```

```python
df_O3.head()
```

```python
df_O3["Date (Local)"] = pd.to_datetime(df_O3["Date (Local)"])
```

```python
daily_O3 = df_O3.groupby("Date (Local)")["Arithmetic Mean"].mean().reset_index()
```

```python
plt.figure(figsize=(12, 6))
```

```python
sns.lineplot(data=daily_O3, x="Date (Local)", y="Arithmetic Mean")
plt.show()
```

```python
plt.show()
```

```python
highest_O3 = daily_O3[daily_O3["Arithmetic Mean"] == daily_O3["Arithmetic Mean"].max()]
highest_O3
```

```python
above_35 = daily_O3[daily_O3["Arithmetic Mean"] > 0.035]
number_of_days = len(above_35)
print("Number of days with O3 concentrations above 35 ppbv:", number_of_days)
```

```python
sns.lineplot(data=daily_O3, x="Date (Local)",y="Arithmetic Mean")

# Add the 35 ppbv threshold line
plt.axhline(y=0.035, linestyle="--", label="35 ppbv")

plt.xlabel("Date")
plt.ylabel("Daily Average O3 (ppm)")
plt.title("Daily Average Hourly O3")

plt.legend()
plt.show()
```
